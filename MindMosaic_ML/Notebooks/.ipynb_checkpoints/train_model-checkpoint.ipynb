```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split , StratifiedKFold,
    cross_val_score , GridSearchCV
)

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)

from xgboost import XGBClassifier

sns.set_style("whitegrid")
```

```python
DATA_PATH = "../Data/game_data.csv"
MODEL_PATH = "../Models/difficulty_model.pkl"

FEATURES = [
    "level" , "card_count" , "shuffle_moves" , "memorize_time",
    "accuracy" , "mistakes" , "reaction_time" , "completion_time",
    "best_streak",
]

TARGET = "target"
RANDOM_STATE = 42
TEST_SIZE = 0.2
cv_folds = 5
```

```python
df = pd.read_csv(DATA_PATH)

print(f"Shape : {df.shape}")
print(f"\nDtypes:\n{df.dtypes}")
print(f"\n Target balance: \n {df[TARGET].value_counts(normalize=True)}")
print(f"\nFeature stats:")
print(df[FEATURES].describe().round(3))
```

```python
plt.figure(figsize = (10 , 8))
sns.heatmap(df[FEATURES + [TARGET]].corr(),
            annot = True,
            fmt = ".2f",
            cmap = "coolwarm",
            center = 0)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
X = df[FEATURES]
y = df[TARGET]

X_train , X_test , y_train , y_test = train_test_split(
    X , y , test_size = TEST_SIZE , random_state = RANDOM_STATE , stratify = y
)

print(f"Train : {X_train.shape} , Test: {X_test.shape}")
```

```python
def evaluate(name , model , X_t , y_t):
    y_pred = model.predict(X_t)
    acc = accuracy_score(y_t , y_pred)
    prec = precision_score(y_t , y_pred , zero_division=0)
    rec = recall_score(y_t , y_pred , zero_division=0)
    f1 = f1_score(y_t , y_pred , zero_division=0)
    cm = confusion_matrix(y_t , y_pred)

    print(f"{name}")
    print(f" Accuracy : {acc:.4f}")
    print(f" Precision : {prec:.4f}")
    print(f" Recall : {rec:.4f}")
    print(f" F1-Score :{f1:.4f}")
    print(f" Confusion matrix:")
    print(f" TN = {cm[0,0]}  FP = {cm[0,1]}")
    print(f" FN={cm[1,0]}  TP = {cm[1,1]}")
    return f1
```

```python
lr_pipe = Pipeline([
    ("scaler" , StandardScaler()),
    ("clf" , LogisticRegression(
        max_iter = 1000,
        random_state = RANDOM_STATE
    )),
])

rf_pipe = Pipeline([
    ("scaler" , StandardScaler()),
    ("clf" , RandomForestClassifier(
        n_estimators = 200,
        max_depth = 10,
        random_state = RANDOM_STATE,
        n_jobs = -1
    )),
])

xgb_pipe = Pipeline([
    ("scaler" , StandardScaler()),
    ("clf" , XGBClassifier(
        n_estimators=200,
        learning_rate = 0.1,
        max_depth=6,
        subsample = 0.8,
        colsample_bytree = 0.8,
        random_state = RANDOM_STATE,
        eval_metric = "logloss",
        verbosity = 0,
    )),
])
```

```python
lr_pipe.fit(X_train , y_train)
lr_f1_base = evaluate("Logistic Regression" , lr_pipe , X_test , y_test)

rf_pipe.fit(X_train , y_train)
rf_f1_base = evaluate("Random Forest" , rf_pipe , X_test , y_test)

xgb_pipe.fit(X_train , y_train)
xgb_f1_base = evaluate("XGBoost" , xgb_pipe , X_test , y_test)
```

```python
cv = StratifiedKFold(n_splits = cv_folds , shuffle = True , random_state = RANDOM_STATE)

lr_cv = cross_val_score(lr_pipe , X_train , y_train , cv=cv , scoring = "f1")
rf_cv = cross_val_score(rf_pipe , X_train , y_train , cv=cv , scoring = "f1")
xgb_cv = cross_val_score(xgb_pipe , X_train , y_train , cv=cv , scoring = "f1")

print(f"LR CV F1 : {lr_cv.mean():.4f} +- {lr_cv.std():.4f}")
print(f"RF CV F1 : {rf_cv.mean():.4f} +- {rf_cv.std():.4f}")
print(f"XGB CV F1 : {xgb_cv.mean():.4f} +- {xgb_cv.std():.4f}")
```

```python
lr_params = {
    "clf__C" : [0.01 , 0.1 , 1 , 10 , 100],
}

lr_grid = GridSearchCV(
    lr_pipe , lr_params,
    cv=cv , scoring="f1" , n_jobs=-1 , refit = True
)

lr_grid.fit(X_train , y_train)

print(f"Best Params : {lr_grid.best_params_}")
print(f"Best CV f1 : {lr_grid.best_score_:.4f}")

lr_f1_tuned = evaluate("LR-tuned" , lr_grid.best_estimator_ , X_test , y_test)

best_lr = lr_grid.best_estimator_ if lr_f1_tuned >= lr_f1_base else lr_pipe
best_lr_f1 = max(lr_f1_tuned , lr_f1_base)
best_lr_label = "LR tuned" if lr_f1_tuned >= lr_f1_base else "LR Baseline"
print(f"\n Best LR :{best_lr_label} (F1 = {best_lr_f1:.4f})")
```

```python
rf_params = {
    "clf__n_estimators" : [100 , 200 , 300],
    "clf__max_depth" : [5 , 10 , 15 , None],
    "clf__min_samples_split" : [2 , 5 , 10],
}

rf_grid = GridSearchCV(
    rf_pipe , rf_params,
    cv=cv , scoring="f1" , n_jobs=-1 , refit=True
)

rf_grid.fit(X_train , y_train)

print(f"Best params : {rf_grid.best_params_}")
print(f"Best CV F1  : {rf_grid.best_score_:.4f}")

rf_f1_tuned = evaluate("RF-tuned" , rf_grid.best_estimator_ , X_test , y_test)

best_rf = rf_grid.best_estimator_ if rf_f1_tuned >= rf_f1_base else rf_pipe
best_rf_f1 = max(rf_f1_tuned , rf_f1_base)
best_rf_label = "RF tuned" if rf_f1_tuned >= rf_f1_base else "RF baseline"
print(f"\n Best RF : {best_rf_label} (F1 = {best_rf_f1:.4f})")
```

```python
xgb_params = {
    "clf__n_estimators" : [100 , 200 , 300],
    "clf__max_depth" : [3 , 5 , 6 , 8],
    "clf__learning_rate" : [0.01 , 0.05 , 0.1 , 0.2],
    "clf__subsample" : [0.7 , 0.8 , 1.0],
}

xgb_grid = GridSearchCV(
    xgb_pipe , xgb_params,
    cv=cv , scoring = "f1" , n_jobs=-1 , refit=True
)

xgb_grid.fit(X_train , y_train)

print(f"Best params : {xgb_grid.best_params_}")
print(f"Best CV F1  : {xgb_grid.best_score_:.4f}")

xgb_f1_tuned = evaluate("XGB-tuned" , xgb_grid.best_estimator_ , X_test , y_test)

best_xgb = xgb_grid.best_estimator_ if xgb_f1_tuned >= xgb_f1_base else xgb_pipe
best_xgb_f1 = max(xgb_f1_tuned , xgb_f1_base)
best_xgb_label = "XGB tuned" if xgb_f1_tuned >= xgb_f1_base else "XGB Basline"
print(f"\n→ Best XGB : {best_xgb_label}  (F1 = {best_xgb_f1:.4f})")
```
